```python
# Task 1 — Averaging Blur
# Load an image, apply a simple averaging blur using OpenCV's blur function.
# Look up the function name yourself — it's a basic box filter.

import cv2
import matplotlib.pyplot as plt

img_brg = cv2.imread("/content/football image.jpg")

img = cv2.cvtColor(img_brg, cv2.COLOR_BGR2RGB)

kernal = (5,5)
avg_blur = cv2.blur(img,kernal)
fig,axes = plt.subplots(1,2,figsize=(14,5))

#original
axes[0].imshow(img)
axes[0].set_title("Original")
axes[0].axis("off")

#Blurred

axes[1].imshow(avg_blur)
axes[1].set_title("Blurred")
axes[1].axis("off")

plt.show()
```

```python
# Task 2 — Gaussian Blur
# Apply Gaussian blur to the same image. Look up the function —
# it takes the kernel size plus one more parameter you'll need to set
#  (look up what it does, don't just default it to 0 blindly... actually,
#   look up what passing 0 does too, it's relevant).


import cv2
import matplotlib.pyplot as plt

img_bgr = cv2.imread("/content/football image.jpg")
img = cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGB)
kernel = (5,5)
guass_blur = cv2.GaussianBlur(img,kernel,0)
fig,axes = plt.subplots(1,2,figsize =(12,5))

#original
axes[0].imshow(img)
axes[0].set_title("Original")
axes[0].axis("off")

#Blurred

axes[1].imshow(guass_blur)
axes[1].set_title("Blurred")
axes[1].axis("off")

plt.show()
```

```python
# Task 3 — Median Blur
# Apply median blur to the same image. Look up the function —
# it takes the image and one parameter (kernel size, but note the format is
# different from the tuple you used before — check what type it expects).

import cv2
import matplotlib.pyplot as plt

img_bgr = cv2.imread("/content/football image.jpg")
img = cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGB)

kernel = 5

Median_blur = cv2.medianBlur(img,kernel)
fig,axes = plt.subplots(1,2,figsize =(12,5))

#original
axes[0].imshow(img)
axes[0].set_title("Original")
axes[0].axis("off")

#Blurred

axes[1].imshow(Median_blur)
axes[1].set_title("Blurred")
axes[1].axis("off")

plt.show()

```

```python
# It smooths while trying to preserve edges (unlike the previous three, which blur edges too).
# Look up cv2.bilateralFilter — it takes the image plus three parameters:
# diameter, sigmaColor, sigmaSpace. Look up what each one controls before you write it.

import cv2
import matplotlib.pyplot as plt

img_bgr = cv2.imread("/content/football image.jpg")
img = cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGB)

bilateral_blur = cv2.bilateralFilter(img,9,75,75)
fig,axes = plt.subplots(1,2,figsize =(12,5))

#original
axes[0].imshow(img)
axes[0].set_title("Original")
axes[0].axis("off")

#Blurred

axes[1].imshow(bilateral_blur)
axes[1].set_title("Blurred")
axes[1].axis("off")

plt.show()
```

```python
# Task 5 — Canny Edge Detection
# Last one for today. Apply Canny edge detection to the image. Look up cv2.Canny —
#  it needs two threshold values. Before you write code: Canny works best on a
#   grayscale image, not color — think about why, and convert accordingly.


import cv2
import matplotlib.pyplot as plt

img_bgr = cv2.imread("/content/football image.jpg")
img = cv2.cvtColor(img_bgr,cv2.COLOR_BGR2GRAY)

canny = cv2.Canny(img,100,150)
fig,axes = plt.subplots(1,2,figsize =(12,5))

#original
axes[0].imshow(img,cmap="gray")
axes[0].set_title("Original")
axes[0].axis("off")

#Blurred

axes[1].imshow(canny,cmap="gray")
axes[1].set_title("Edge detected")
axes[1].axis("off")

plt.show()

```
